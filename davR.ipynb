```r

# Load library
library(lattice)

# Load dataset
data(iris)

# 1. Box Plot
bwplot(Species ~ Petal.Length, data = iris, main = "Boxplot of Petal Length by Species")

# 2. Histogram
histogram(~ Petal.Length, data = iris, main = "Histogram of Petal Length")

# 3. Density Plot
densityplot(~ Petal.Length, data = iris, main = "Density Plot of Petal Length")

# 4. Scatter Plot
xyplot(Petal.Width ~ Petal.Length, data = iris, main = "Scatter Plot")

# 5. Scatter Plot with Grouping
xyplot(Petal.Width ~ Petal.Length | Species, data = iris, main = "Scatter Plot by Species")

# 6. Box Plot (Grouped)
bwplot(Species ~ Sepal.Length, data = iris, main = "Boxplot of Sepal Length by Species")
```

```r

# Load dataset (if not already loaded)
data(iris)

# 1. Box Plot using base R
boxplot(Petal.Length ~ Species, data = iris,
        main = "Boxplot of Petal Length by Species (Base R)",
        xlab = "Species", ylab = "Petal Length")

# 2. Histogram using base R
hist(iris$Petal.Length,
     main = "Histogram of Petal Length (Base R)",
     xlab = "Petal Length", col = "lightblue", border = "black")

# 3. Density Plot using base R
plot(density(iris$Petal.Length),
     main = "Density Plot of Petal Length (Base R)",
     xlab = "Petal Length")

# 4. Scatter Plot using base R
plot(iris$Petal.Length, iris$Petal.Width,
     main = "Scatter Plot of Petal Length vs. Width (Base R)",
     xlab = "Petal Length", ylab = "Petal Width",
     col = iris$Species, pch = 19)

# Add legend for scatter plot
legend("topleft", legend = levels(iris$Species),
       col = 1:length(levels(iris$Species)), pch = 19)
```
